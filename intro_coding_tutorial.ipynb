# 1. What is a variable?

A variable is a name you stick onto a value so you can use it again later.

That's it. The `=` sign here doesn't mean "equals" the way it does in math it means "take the thing on the right, and assign it to the name on the left."

````python
firing_rate = 12.4

print(firing_rate)
````

Now `firing_rate` is a name we can use anywhere. The computer will look up what's stored under that name and use it.

````python
print(firing_rate * 2)
print(firing_rate + 100)
````

Values come in different types, and the type decides what you're allowed to do with them.

````python
n_channels   = 8          # an integer (int): a whole number
firing_rate  = 12.4        # a float: a number with a decimal point
subject_name = "fly_07"  # a string (str): text, always in quotes
eyes_open    = True        # a boolean (bool): True or False, nothing else

# type() tells you which is which
print(type(n_channels))
print(type(firing_rate))
print(type(subject_name))
print(type(eyes_open))
````

Type matters a lot! `+` means "add" for numbers but "stick together" for strings

````python
print(2 + 3)             # 5
print("2" + "3")         # "23" - these are text, so they get glued together!
````

A variable keeps whatever you put in it most recently

````python
count = 1
print("count is now:", count)
````

````python
count = count + 10     # take whatever count is, add 10, put it back under the same name
print("count is now:", count)
````

# 2. Holding lots of values at once

One number at a time won't get us far. A **list** holds many values, in order, in a single variable. You write it with square brackets.

````python
firing_rates = [12.4, 8.1, 30.0, 5.5, 22.7]

print(firing_rates)
print("how many values:", len(firing_rates))
````

To pull one value out, you give its position in square brackets. **Counting starts at 0**, not 1 in Python. In MATLAB, counting start at 1. Confusing and annoying!

So in a list of 5 things, the positions are 0, 1, 2, 3, 4. There is no position 5.

````python
print("first value: ", firing_rates[0])
print("second value:", firing_rates[1])
print("last value:  ", firing_rates[4])
print("also last:   ", firing_rates[-1])   # negative counts backwards from the end
````

You can also take a _slice_ - a chunk of the list. `firing_rates[1:4]` means "from position 1, up to but not including position 4."

````python
print(firing_rates[1:4])   # positions 1, 2, 3
print(firing_rates[:3])    # from the start up to position 3
print(firing_rates[2:])    # from position 2 to the end
````

A **loop** does the same thing to every item in turn. Read it out loud- "for each rate in firing rates, print the rate"

````python
for rate in firing_rates:
    print("this recording had rate:", rate)
````

The indented part is what gets repeated. Python uses indentation to decide what's inside the loop and what's outside it

# 3. Arrays, and why your data has a _shape_

Lists are fine, but for real data everyone uses **numpy arrays** instead. An array looks a lot like a list, with one enormous advantage: it does math on every element at once, without you writing a loop.

````python
import numpy as np

rates_list  = [12.4, 8.1, 30.0, 5.5, 22.7]
rates_array = np.array(rates_list)     # same numbers, different container

print("array:", rates_array)
print("doubled, no loop needed:", rates_array * 2)
print("mean:", rates_array.mean())
````

Arrays can also have rows and columns, which is how essentially all data gets stored. Every array has a `.shape`, which tells you how big it is in each direction.

````python
# 4 recordings, 3 measurements each
data = np.array([
    [12.4, 0.8, 1.1],
    [ 8.1, 0.3, 0.9],
    [30.0, 1.5, 2.2],
    [ 5.5, 0.2, 0.4],
])

print("shape:", data.shape)     # (rows, columns) = (4, 3)
print("number of rows:", data.shape[0])
print("number of columns:", data.shape[1])
````

````python
print("first row (recording 0):", data[0])
print("first column (measurement 0, all recordings):", data[:, 0])
#                                                          ^ the : means "all rows"
print("one single value, row 2 column 1:", data[2, 1])
````

# 4. Functions: giving a name to a set of steps

A variable gives a name to a value. A **function** gives a name to a set of steps, so you can do them again without retyping them.

`def` starts the definition, the bit in brackets is what you hand it, and `return` is what it gives back.

````python
def average(values):
    total = sum(values)
    return total / len(values)


print(average([12.4, 8.1, 30.0]))
print(average([1, 2, 3, 4, 5]))
````

The value you hand in is called an _argument_.

Functions can take more than one argument, and can have defaults.

````python
def is_above(value, threshold=10):
    return value > threshold


print(is_above(12.4))          # uses the default threshold of 10
print(is_above(12.4, 20))      # or you can say what threshold you want
````

### ⭐ If you're ahead: a class is a function with memory

A **class** bundles some data together with the functions that use that data, so you don't have to keep passing the same things around.

````python
class Recording:
    def __init__(self, rates):
        self.rates = rates            # data the object carries around

    def average(self):                # a function that uses that data
        return sum(self.rates) / len(self.rates)


r = Recording([12.4, 8.1, 30.0])
print(r.rates)
print(r.average())
````

# 7. Data wrangling with pandas

So far, we have used lists and NumPy arrays to store data. But real datasets often contain several different kinds of information: subject names, experimental conditions, measurements, and labels.

For this kind of **tabular data** (data organized into rows and columns), Python commonly uses a package called **pandas**. A pandas table is called a **DataFrame**. Each row is usually one observation, and each column is one variable.

````python
import pandas as pd

recordings = pd.DataFrame({
    "subject": ["fly_01", "fly_02", "fly_03", "fly_04", "fly_05"],
    "condition": ["eyes open", "eyes closed", "eyes open", "eyes closed", "eyes open"],
    "firing_rate": [12.4, 8.1, 30.0, 5.5, 22.7]
})

recordings
````

A DataFrame has a `.shape`, just like a NumPy array. You can also use `.head()` to look at the first few rows and `.columns` to see the column names.

````python
print("shape:", recordings.shape)
print("columns:", recordings.columns)
recordings.head()
````

You select a column by writing its name in square brackets. Then you can calculate things like its mean or maximum.

````python
print(recordings["firing_rate"])
print("mean firing rate:", recordings["firing_rate"].mean())
print("maximum firing rate:", recordings["firing_rate"].max())
````

### Filtering rows

Often, you only want the rows that meet some condition. This works like filtering a NumPy array, except pandas keeps the entire row.

````python
# Keep only recordings with firing rates above 10
recordings[recordings["firing_rate"] > 10]
````

````python
# Keep only recordings from the eyes-open condition
recordings[recordings["condition"] == "eyes open"]
````

You can also make a new column from the columns you already have. Here, each row gets `True` or `False` depending on whether its firing rate is above 10.

````python
recordings["high_firing"] = recordings["firing_rate"] > 10
recordings
````
